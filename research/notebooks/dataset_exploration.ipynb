## Correlation Table of Numerical Attributes

A correlation table is produced that is used for the report.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select numerical columns
numerical_cols = [
    'Previous qualification (grade)',
    'Admission grade',
    'Age at enrollment',
    'Curricular units 1st sem (credited)',
    'Curricular units 1st sem (enrolled)',
    'Curricular units 1st sem (evaluations)',
    'Curricular units 1st sem (approved)',
    'Curricular units 1st sem (grade)',
    'Curricular units 1st sem (without evaluations)',
    'Curricular units 2nd sem (credited)',
    'Curricular units 2nd sem (enrolled)',
    'Curricular units 2nd sem (evaluations)',
    'Curricular units 2nd sem (approved)',
    'Curricular units 2nd sem (grade)',
    'Curricular units 2nd sem (without evaluations)',
    'Unemployment rate',
    'Inflation rate',
    'GDP'
]

# Compute correlation matrix
correlation_matrix = df[numerical_cols].corr()

# Create figure for correlation heatmap
plt.figure(figsize=(14, 10))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0,
            cbar_kws={'label': 'Correlation Coefficient'}, square=True)
plt.title('Correlation Matrix of Numerical Attributes', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Features', fontsize=12, fontweight='bold')
plt.ylabel('Features', fontsize=12, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Histograms for Numerical Features

```python
# Create subplots for histograms
fig, axes = plt.subplots(6, 3, figsize=(16, 18))
axes = axes.flatten()

# Plot histograms for each numerical column
for idx, col in enumerate(numerical_cols):
    axes[idx].hist(df[col].dropna(), bins=30, color='steelblue', edgecolor='black', alpha=0.7)
    axes[idx].set_title(f'Distribution of {col}', fontsize=10, fontweight='bold')
    axes[idx].set_xlabel('Value', fontsize=9)
    axes[idx].set_ylabel('Frequency', fontsize=9)
    axes[idx].grid(axis='y', alpha=0.3, linestyle='--')

# Hide the last unused subplot
axes[-1].set_visible(False)

plt.suptitle('Histograms of Numerical Features', fontsize=16, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()
```

## Bar Charts for Non-Numerical Features

```python
# Select non-numerical (categorical) columns
categorical_cols = [
    'Marital Status',
    'Application mode',
    'Application order',
    'Course',
    'Daytime/evening attendance',
    'Previous qualification',
    'Nationality',
    "Mother's qualification",
    "Father's qualification",
    "Mother's occupation",
    "Father's occupation",
    'Displaced',
    'Educational special needs',
    'Debtor',
    'Tuition fees up to date',
    'Gender',
    'Scholarship holder',
    'International',
    'Target'
]

# Create subplots for bar charts
fig, axes = plt.subplots(5, 4, figsize=(16, 15))
axes = axes.flatten()

# Plot bar charts for each categorical column
for idx, col in enumerate(categorical_cols):
    value_counts = df[col].value_counts().sort_index()
    axes[idx].bar(value_counts.index.astype(str), value_counts.values, color='coral', edgecolor='black', alpha=0.7)
    axes[idx].set_title(f'{col}', fontsize=10, fontweight='bold')
    axes[idx].set_xlabel('Category', fontsize=9)
    axes[idx].set_ylabel('Count', fontsize=9)
    axes[idx].grid(axis='y', alpha=0.3, linestyle='--')
    axes[idx].tick_params(axis='x', rotation=45)

# Hide the last unused subplot
axes[-1].set_visible(False)

plt.suptitle('Bar Charts of Non-Numerical Features', fontsize=16, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()
```

## Cleaning operations

* Remove GDP, inflation and unemployment rate as their meaning are unambigous as an attribute describing a student.
* Investigate the curricular unit distributions accross different semesters. If they have very similar distributions, remove one of the semesters.
* Replace 'Target' column name with 'Graduation status'.
* Encode Graduation status values

```python
# Remove inflation rate, gdp, and unemployment rate from df
df = df.drop(columns=['Inflation rate', 'GDP', 'Unemployment rate'])
```

```python
# Rename Target column to Graduation status
df.rename(columns={'Target': 'Graduation status'}, inplace=True)
```

```python
# Check unique values in Graduation status column
df['Graduation status'].unique()
```

```python
# Encode Dropout with 0, Enrolled with 1, and Graduate with 2
df['Graduation status'] = df['Graduation status'].map({'Dropout': 0, 'Enrolled': 1, 'Graduate': 2})
df['Graduation status'].unique()
```

```python

curricular_unit_semester_pairs = [
    ("Curricular units 1st sem (credited)", "Curricular units 2nd sem (credited)"),
    ("Curricular units 1st sem (enrolled)", "Curricular units 2nd sem (enrolled)"),
    ("Curricular units 1st sem (evaluations)", "Curricular units 2nd sem (evaluations)"),
    ("Curricular units 1st sem (approved)", "Curricular units 2nd sem (approved)"),
    ("Curricular units 1st sem (grade)", "Curricular units 2nd sem (grade)"),
    ("Curricular units 1st sem (without evaluations)", "Curricular units 2nd sem (without evaluations)")
]

# Build overlaying histograms for each pair in curricular_unit_semester_pairs
for first_sem, second_sem in curricular_unit_semester_pairs:
    plt.figure(figsize=(10, 6))
    sns.histplot(df[first_sem], color='blue', label='1st Sem', kde=True, stat="density", alpha=0.5)
    sns.histplot(df[second_sem], color='red', label='2nd Sem', kde=True, stat="density", alpha=0.5)
    plt.title(f'Overlaying Histograms: {first_sem} vs {second_sem}', fontsize=14, fontweight='bold')
    plt.xlabel('Value', fontsize=12, fontweight='bold')
    plt.ylabel('Density', fontsize=12, fontweight='bold')
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
# Remove curricular units of the second semester as they are very similar to the first semester
df = df.drop(columns=[
    "Curricular units 2nd sem (credited)",
    "Curricular units 2nd sem (enrolled)",
    "Curricular units 2nd sem (evaluations)",
    "Curricular units 2nd sem (approved)",
    "Curricular units 2nd sem (grade)",
    "Curricular units 2nd sem (without evaluations)"
])

# Remove "1st sem" in curricular unit column names
df = df.rename(columns=lambda x: x.replace("Curricular units 1st sem ", "Curricular units "))
```
