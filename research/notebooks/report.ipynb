## Dataset (Task 1)

The [**Students dropout and academic success**](https://archive.ics.uci.edu/dataset/697/predict+students+dropout+and+academic+success) dataset originates from the educational domain and provide insights in students' academic performances, dropout and graduation rates, demographics, along with their educational and socio-economic situations. The dataset is supported by the SATDAP - Capacitação da Administração Pública program and was collected from the *UCI machine learning repository* where it was used to train and test a classification model to predict student dropout and academic success. The dataset contains more than 4000 students that are described by 37 different attributes with no missing values. The dataset has a tabular format with rows and columns representing individual students and student attributes respectively. The attributes describe varying aspects of a student, including martial status, course enrollment, when they attended class, education level, nationality, parental education level, and grades. The attribute types include numerical, nominal, ordinal, and binary types. The attribtues and their types and values are outlined in detail in the table below.

### Justification of suitability (Task 1)

The dataset is suitable for projection for several reasons, the major being the existence of many (>10) numerical dimensions, including admission grades and varying curricular units, which cannot be represented in three dimensions or lower. Further, the many nominal features allows us to encode the points in the reduced space based on these values, making the potential analyses multivariate.
